```python
from lifespan_calendar import SurvivalCalendar, write_svg
```

```python
import numpy as np
```

```python
num_weeks_alive = 1_319
survival_probs: list[int] = [1] * num_weeks_alive + list(np.linspace(start=1.0, stop=0, num=3361))
calendar_generator = SurvivalCalendar(num_weeks_survived=num_weeks_alive, survival_probabilities=survival_probs)
```

```python
calendar_xml: str = calendar_generator.to_web_svg(radius=6, stroke_width=1)
```

```python
write_svg(calendar_xml, 'examples/run2.svg')
```

```python
calendar_generator.to_png("run2.png", radius=6, stroke_width=1, dpi=300, supersample=2)
```
